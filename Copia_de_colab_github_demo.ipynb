```
import pandas as pd
import matplotlib.pyplot as plt

# Cargar datos desde archivos CSV
pressure_and_velocity = pd.read_csv('Pressure_and_velocity.csv', delimiter='\t', header=0, na_values='')
acceleration = pd.read_csv('Acceleration.csv', delimiter='\t', header=0, na_values='')

# Extraer columnas
th = pressure_and_velocity['Time (s)']
h = pressure_and_velocity['Altitude (m)']
p = pressure_and_velocity['Pressure (hPa)']

tv = pressure_and_velocity['Time (velocity) (s)']
yv = pressure_and_velocity['Velocity (m/s)']


ta = acceleration['Time (s)']
A = acceleration['Acceleration (m/s²)']


# Convert relevant columns to numeric, coercing errors
th = pd.to_numeric(th, errors='coerce')
h = pd.to_numeric(h, errors='coerce')
p = pd.to_numeric(p, errors='coerce')
tv = pd.to_numeric(tv, errors='coerce')
yv = pd.to_numeric(yv, errors='coerce')
ta = pd.to_numeric(ta, errors='coerce')
A = pd.to_numeric(A, errors='coerce')


# Crear figura y subplots
fig, axs = plt.subplots(4, 1, figsize=(10, 12))

axs[0].plot(th, h, color='#AF172E')
axs[0].set_xlabel('Time [s]')
axs[0].set_ylabel('h [m]')
axs[0].grid(True)

axs[1].plot(tv, yv)
axs[1].set_xlabel('Time [s]')
axs[1].set_ylabel('Vel.Vert. [m/s]')
axs[1].grid(True)

axs[2].plot(ta, A, color='#0DA50D')
axs[2].set_xlabel('Time [s]')
axs[2].set_ylabel('Acc.Z [m/s^2]')
axs[2].grid(True)

axs[3].plot(th, p, color='#D16A12')
axs[3].set_xlabel('Time [s]')
axs[3].set_ylabel('P [hPa]')
axs[3].grid(True)

# Ajustar espacios
plt.tight_layout()

# Guardar figura
plt.savefig('Grafica2.png')
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt
# Cargar datos desde archivos CSV
accel = pd.read_csv('Raw Data.csv', delimiter='\t', header=0, na_values='', decimal=',')

# Extraer columnas
t = accel['Time (s)']
x = accel['Linear Acceleration x (m/s^2)']
y = accel['Linear Acceleration y (m/s^2)']
z = accel['Linear Acceleration z (m/s^2)']
aabs = accel['Absolute acceleration (m/s^2)']



# Convert relevant columns to numeric, coercing errors
t = pd.to_numeric(t, errors='coerce')
x = pd.to_numeric(x, errors='coerce')
y = pd.to_numeric(y, errors='coerce')
z = pd.to_numeric(z, errors='coerce')
aabs = pd.to_numeric(aabs, errors='coerce')

# Crear figura y subplots
fig, axs = plt.subplots(4, 1, figsize=(10, 12))

axs[0].plot(t, x, color='#AF172E')
axs[0].set_xlabel('Time [s]')
axs[0].set_ylabel('Acc.X [m/s^2]')
axs[0].grid(True)

axs[1].plot(t, y)
axs[1].set_xlabel('Time [s]')
axs[1].set_ylabel('Acc.Y [m/s^2]')
axs[1].grid(True)

axs[2].plot(t, z, color='#0DA50D')
axs[2].set_xlabel('Time [s]')
axs[2].set_ylabel('Acc.Z [m/s^2]')
axs[2].grid(True)

axs[3].plot(t, aabs, color='#D16A12')
axs[3].set_xlabel('Time [s]')
axs[3].set_ylabel('Abs.Acc. [m/s^2]')
axs[3].grid(True)

# Ajustar espacios
plt.tight_layout()

# Guardar figura
plt.savefig('Grafica.png')
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt
import os

Echo = pd.read_csv('Echo location.csv', delimiter='\t', header=0, na_values='')
Speed = pd.read_csv('Speed of sound.csv', delimiter='\t', header=0, na_values='')
Time = pd.read_csv('Time.csv',delimiter='\t', header=0, na_values='')

# Extraer columnas
delay = Time['Delay (s)']
Adelay = Time['Crosscorrelation (a.u.)']

Distance = Echo['Distance (cm)']
ADistance = Echo['Crosscorrelation (a.u.)']

vel = Speed['Speed of sound (m/s)']
Avel = Speed['Crosscorrelation (a.u.)']

    # Convert relevant columns to numeric, coercing errors
delay = pd.to_numeric(delay, errors='coerce')
Adelay = pd.to_numeric(Adelay, errors='coerce')
Distance = pd.to_numeric(Distance, errors='coerce')
ADistance = pd.to_numeric(ADistance, errors='coerce')
vel = pd.to_numeric(vel, errors='coerce')
Avel = pd.to_numeric(Avel, errors='coerce')

# Crear figura y subplots
fig, axs = plt.subplots(3, 1, figsize=(10, 12))

axs[0].plot(delay, Adelay, color='#AF172E')
axs[0].set_xlabel('Retardo [s]')
axs[0].set_ylabel('Amplitud [a.u]')
axs[0].grid(True)

axs[1].plot(Distance, ADistance)
axs[1].set_xlabel('Distancia [cm]')
axs[1].set_ylabel('Amplitud [a.u]')
axs[1].grid(True)

axs[2].plot(vel, Avel, color='#0DA50D')
axs[2].set_xlabel('Velocidad del sonido [m/s]')
axs[2].set_ylabel('Amplitud [a.u]')
axs[2].grid(True)

# Ajustar espacios
plt.tight_layout()
# Guardar figura
plt.savefig('Grafica2.png')
plt.show()
```
